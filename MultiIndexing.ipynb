```python
import pandas as pd
import numpy as np
import random
```

## Creating and Understanding MultiIndex

```python
df = pd.read_csv('athlete_events.csv')
df.head()
```

### Setting the index on `Year` and `Season`

```python
df= df.set_index(['Year','Season'])
```

### Setting the index on `Year`, `Season` and `Sport`

```python
df = df.set_index(['Year','Season','Sport'])
df = df.sort_index()
df.head()
```

### What if I want information about all people who participated in Archery across all Years and Season ??

```python
df.xs('Archery', level='Sport').head()
```

### I want to find average Weight of all participants in Football Sport¶

```python
round(df.xs('Football', level='Sport')['Weight'].mean(),2)
```

### I want to find the average weight of all participants however they should be grouped by their gender

```python
df.xs('Football', level='Sport')[['Sex','Weight']].groupby('Sex').mean()
```

### I want to Specify more than one things; For eg: what if I want info about all archer in the Year 2000

```python
df.xs((2000, 'Archery'), level=['Year', 'Sport']).head()
```

### Select the rows where the `Year is 2000`

```python
df.loc[2000].head()
```

### Select the rows where the `Year is 2000` and `Season is Summer`



```python
df.loc[2000,'Summer'].head()
```

### Select the rows where `Year is 2000`, `Season is Summer` and `Sport is Basketball`

### I want information about all people who participated in Archery across all Years and Season ??



```python
# df.xs(level, axis=0, level=None, drop_level=True)
# level will be the category in that level
# level= will be the column name
df.xs('Archery',level='Sport').head()
```

### Select all rows where the Year is greater than or equal to 2000.

```python
df[df.index.get_level_values(0)>=2000]
```

### Select all people who have participated in 'Handball' since the Year 2013

```python
df[(df.index.get_level_values(0) >=2013) & (df.index.get_level_values(2) == 'Handball')]
```

### How Many Category of sports exists

### Swap the levels of the Sports with Season

```python
df.swaplevel('Sport','Season').head()
# To make this change for long df = df.swaplevel('Sport','Season')
```

### Sort the DataFrame by the Sports level in descending order.

```python
df=df.sort_index(level='Sport',ascending=False)
df.head()
```

### Get the average weight for all sports and all season

```python
grouped = df.groupby(level=['Sport','Season'])['Weight'].mean()
```

### Get the count of the gender for all category of sports

```python
grouped=df.groupby(['Sport','Sex']).count()
grouped.head()
```

### Get the average heights of the gender for all category of sports for all Teams

```python
df.groupby(['Team','Sex','Sport'])['Height'].mean()
```
